Create your first GDS

```python
import nazca as nd

#The code to generate a rectangle of length 10um in the cell
with nd.Cell(name='cellname') as cell_variable:
  nd.strt(length=10).put()

#Placing another rectangle 10um on the xaxis and 20um on the y axis in respect to the first one
  nd.strt(length=10).put(10, 20)

  nd.strt(length=15, width=3.5).put(0, 30, 45)
  nd.strt(length=15, layer=3).put(0,-20)

cell_variable.put()

nd.export_gds(filename="/content/GDS Output/first gds.gds")
```

Creating an Emoji

```python
import nazca as nd

with nd.Cell(name='cellname') as cell_variable:
  nd.strt(length=0.02, width=0.005).put()
  nd.strt(length=0.01, width=0.001).put(0.1,-0.008)
  nd.strt(length=0.02, width=0.005).put(0.2,0)

cell_variable.put()
nd.export_gds(filename="/content/GDS Output/first gds.gds")


```

Creating the number 3

```python
import nazca as nd

with nd.Cell(name='cellname') as cell_variable:
  nd.strt(length=15).put()
  nd.strt(length=15).put(0,-5)
  nd.strt(length=15).put(0,-10)
  nd.strt(length=2, width=11).put(15,-5)

cell_variable.put()
nd.export_gds(filename="/content/GDS Output/first gds.gds")
```

```python

with nd.Cell(name='Cell 2') as C:
  line = nd.strt(length=15).put()
  line2 = nd.strt(length=30, layer=3).put(line)#using line command places the rect next to each other

  bend1 = nd.bend(radius=15, layer=6).put(line2)

  nd.bend(radius=15, angle=45).put(bend1)
C.put()

nd.export_gds(filename="/content/GDS Output/first gds.gds")
```

Creating letter o

```python
with nd.Cell(name='Cell 2') as C:


  bend = nd.bend(radius=15).put()
  bend1 = nd.bend(radius=15).put(bend)
  bend2 = nd.bend(radius=15).put(bend1)
  bend = nd.bend(radius=15).put(bend2)


C.put()

nd.export_gds(filename="/content/GDS Output/first gds.gds")
```

creating letter s

```python
with nd.Cell(name='Cell 2') as C:


  bend = nd.bend(radius=15, angle=180).put()
  bend1 = nd.bend(radius=15, angle=-180).put(bend)



C.put()

nd.export_gds(filename="/content/GDS Output/first gds.gds")
```

creating +

```python
with nd.Cell(name='Cell 2') as C:
  line = nd.strt(length=15).put()
  line2 = nd.strt(length=1, width=15).put(7,0)


C.put()

nd.export_gds(filename="/content/GDS Output/first gds.gds")
```

Alternative best way of making +

```python
with nd.Cell(name='Cell 2') as C:
  plus_size=15 #This allows us to control the width and length of the lines

  nd.strt(length=15).put()
  nd.strt(length=15).put(plus_size/2, -plus_size/2, 90)


C.put()

nd.export_gds(filename="/content/GDS Output/first gds.gds")
```
